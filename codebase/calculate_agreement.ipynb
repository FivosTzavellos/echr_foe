File that calculates Kappa Cohen's agreement score between annotators for the task of dataset validation

```python
import pandas as pd
from sklearn.metrics import cohen_kappa_score
```

```python
file_path = '../dataset/new_version_echr_classification_annotation.xlsx'  

sheet_names = ['Annotator_1', 'Annotator_2']

data = {}
for sheet in sheet_names:
    df = pd.read_excel(file_path, sheet_name=sheet)
    data[sheet] = df['Manual_Label']  

data
```

```python
label_mapping = {
    'violation_of_article_10': 0,
    'no_violation_of_article_10': 1,
    'other': 2
}

for key in data:
    data[key] = data[key].map(label_mapping)

kappa_annotators = cohen_kappa_score(data['Annotator_1'], data['Annotator_2'])
print(f"Cohen's Kappa between Annotator 1 and Annotator 2: {kappa_annotators:.2f}")
```
